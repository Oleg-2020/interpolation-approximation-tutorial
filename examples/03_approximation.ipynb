# Example 3: Function Approximation with Different Basis

Demonstrates least squares approximation using Chebyshev, Legendre, and power bases.

```python
import numpy as np
import matplotlib.pyplot as plt
from mathdoc import DataLoader, Approximator, ErrorAnalyzer
```

## 3.1 Generate Noisy Data

```python
loader = DataLoader(random_seed=42)
x, y = loader.load_synthetic(n_points=60, noise=0.08, function='sine')

# Create exact function for reference
x_exact = np.linspace(0, 1, 200)
y_exact = np.sin(2*np.pi*x_exact)

plt.figure(figsize=(10, 5))
plt.scatter(x, y, alpha=0.5, s=40, label='Noisy data')
plt.plot(x_exact, y_exact, 'g--', linewidth=2, label='Exact function')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Data for Approximation', fontsize=12, fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 3.2 Approximation with Chebyshev Basis

```python
# Fit with Chebyshev basis
approx_cheb = Approximator(x, y, basis_type='chebyshev', degree=8)
coeffs_cheb = approx_cheb.fit()
y_approx_cheb = approx_cheb.evaluate(x_exact)

# Error analysis
error_analyzer = ErrorAnalyzer(y_exact, y_approx_cheb)
error_summary = error_analyzer.summary(y_approx_cheb)

print('Chebyshev Approximation (degree 8):')
print(f'  L2 error norm: {error_summary["l2_norm"]:.6f}')
print(f'  Max error: {error_summary["l_inf_norm"]:.6f}')
print(f'  Relative error: {error_summary["relative_error"]:.6f}')
print(f'  Coefficients: {coeffs_cheb}')
```

## 3.3 Approximation with Legendre Basis

```python
# Fit with Legendre basis
approx_leg = Approximator(x, y, basis_type='legendre', degree=8)
coeffs_leg = approx_leg.fit()
y_approx_leg = approx_leg.evaluate(x_exact)

error_analyzer_leg = ErrorAnalyzer(y_exact, y_approx_leg)
error_summary_leg = error_analyzer_leg.summary(y_approx_leg)

print('Legendre Approximation (degree 8):')
print(f'  L2 error norm: {error_summary_leg["l2_norm"]:.6f}')
print(f'  Max error: {error_summary_leg["l_inf_norm"]:.6f}')
print(f'  Relative error: {error_summary_leg["relative_error"]:.6f}')
```

## 3.4 Approximation with Power Basis

```python
# Fit with power basis
approx_pow = Approximator(x, y, basis_type='power', degree=8)
coeffs_pow = approx_pow.fit()
y_approx_pow = approx_pow.evaluate(x_exact)

error_analyzer_pow = ErrorAnalyzer(y_exact, y_approx_pow)
error_summary_pow = error_analyzer_pow.summary(y_approx_pow)

print('Power Basis Approximation (degree 8):')
print(f'  L2 error norm: {error_summary_pow["l2_norm"]:.6f}')
print(f'  Max error: {error_summary_pow["l_inf_norm"]:.6f}')
print(f'  Relative error: {error_summary_pow["relative_error"]:.6f}')
```

## 3.5 Comparison and Error Analysis

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left: approximations
axes[0].scatter(x, y, alpha=0.4, s=30, label='Data', color='black')
axes[0].plot(x_exact, y_exact, 'g--', linewidth=2, label='Exact', alpha=0.7)
axes[0].plot(x_exact, y_approx_cheb, 'b-', linewidth=1.5, label='Chebyshev')
axes[0].plot(x_exact, y_approx_leg, 'r-', linewidth=1.5, label='Legendre', alpha=0.7)
axes[0].plot(x_exact, y_approx_pow, 'orange', linewidth=1.5, label='Power', alpha=0.7)
axes[0].set_xlabel('x')
axes[0].set_ylabel('y')
axes[0].set_title('Approximations Comparison', fontsize=12, fontweight='bold')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Right: errors
error_cheb = y_exact - y_approx_cheb
error_leg = y_exact - y_approx_leg
error_pow = y_exact - y_approx_pow

axes[1].plot(x_exact, error_cheb, 'b-', linewidth=1.5, label='Chebyshev')
axes[1].plot(x_exact, error_leg, 'r-', linewidth=1.5, label='Legendre', alpha=0.7)
axes[1].plot(x_exact, error_pow, 'orange', linewidth=1.5, label='Power', alpha=0.7)
axes[1].axhline(0, color='k', linestyle='--', linewidth=0.5)
axes[1].set_xlabel('x')
axes[1].set_ylabel('Error')
axes[1].set_title('Approximation Errors', fontsize=12, fontweight='bold')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 3.6 Convergence Analysis

```python
# Test convergence with increasing degree
degrees = range(1, 12)
errors_cheb = []
errors_leg = []
errors_pow = []

for d in degrees:
    # Chebyshev
    app = Approximator(x, y, basis_type='chebyshev', degree=d)
    app.fit()
    y_app = app.evaluate(x_exact)
    errors_cheb.append(ErrorAnalyzer(y_exact, y_app).l2_norm(y_app))
    
    # Legendre
    app = Approximator(x, y, basis_type='legendre', degree=d)
    app.fit()
    y_app = app.evaluate(x_exact)
    errors_leg.append(ErrorAnalyzer(y_exact, y_app).l2_norm(y_app))
    
    # Power
    app = Approximator(x, y, basis_type='power', degree=d)
    app.fit()
    y_app = app.evaluate(x_exact)
    errors_pow.append(ErrorAnalyzer(y_exact, y_app).l2_norm(y_app))

plt.figure(figsize=(10, 6))
plt.semilogy(degrees, errors_cheb, 'o-', linewidth=2, markersize=6, label='Chebyshev')
plt.semilogy(degrees, errors_leg, 's-', linewidth=2, markersize=6, label='Legendre')
plt.semilogy(degrees, errors_pow, '^-', linewidth=2, markersize=6, label='Power')
plt.xlabel('Polynomial Degree', fontsize=11)
plt.ylabel('L2 Error (log scale)', fontsize=11)
plt.title('Convergence Analysis', fontsize=12, fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3, which='both')
plt.show()
```

## Key Observations

- **Chebyshev basis** typically converges fastest (optimal approximation property)
- **Legendre basis** is orthogonal and provides good performance
- **Power basis** is simpler but can suffer from ill-conditioning at high degrees
