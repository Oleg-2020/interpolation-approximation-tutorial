# Example 1: Data Loading and Preprocessing

This notebook demonstrates how to load, validate, and preprocess data for mathematical analysis.

```python
import numpy as np
import matplotlib.pyplot as plt
from mathdoc import DataLoader
```

## 1.1 Generate Synthetic Data

Create synthetic data from mathematical functions.

```python
loader = DataLoader(random_seed=42)

# Generate sine wave data
x_sine, y_sine = loader.load_synthetic(n_points=100, noise=0.05, function='sine')

# Generate polynomial data
x_poly, y_poly = loader.load_synthetic(n_points=50, noise=0.02, function='polynomial')

print(f'Sine data shape: x={x_sine.shape}, y={y_sine.shape}')
print(f'Polynomial data shape: x={x_poly.shape}, y={y_poly.shape}')
```

## 1.2 Visualize Raw Data

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

# Plot sine data
axes[0].scatter(x_sine, y_sine, alpha=0.6, s=30)
axes[0].set_title('Sine Function with Noise', fontsize=12, fontweight='bold')
axes[0].set_xlabel('x')
axes[0].set_ylabel('y')
axes[0].grid(True, alpha=0.3)

# Plot polynomial data
axes[1].scatter(x_poly, y_poly, alpha=0.6, s=30, color='red')
axes[1].set_title('Polynomial with Noise', fontsize=12, fontweight='bold')
axes[1].set_xlabel('x')
axes[1].set_ylabel('y')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 1.3 Remove Outliers

Detect and remove anomalous points.

```python
# Add some outliers to sine data
y_outliers = y_sine.copy()
outlier_indices = [10, 50, 80]
y_outliers[outlier_indices] += 5

# Remove outliers
x_clean, y_clean = loader.remove_outliers(x_sine, y_outliers, threshold=2.0)

print(f'Original data points: {len(y_outliers)}')
print(f'Cleaned data points: {len(y_clean)}')
print(f'Removed {len(y_outliers) - len(y_clean)} outliers')
```

## 1.4 Normalize Data

Normalize using different methods.

```python
# Min-Max normalization
y_minmax = loader.normalize(y_sine, method='minmax')

# Z-score normalization
y_zscore = loader.normalize(y_sine, method='zscore')

print(f'Original data range: [{y_sine.min():.3f}, {y_sine.max():.3f}]')
print(f'Min-Max normalized range: [{y_minmax.min():.3f}, {y_minmax.max():.3f}]')
print(f'Z-score mean: {y_zscore.mean():.6f}, std: {y_zscore.std():.6f}')
```

## 1.5 Comparison

Visualize original, noisy, and cleaned data.

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Original
axes[0].scatter(x_sine, y_sine, alpha=0.6)
axes[0].set_title('Original (Clean)', fontsize=12)
axes[0].grid(True, alpha=0.3)

# With outliers
axes[1].scatter(x_sine, y_outliers, alpha=0.6)
axes[1].scatter(x_sine[outlier_indices], y_outliers[outlier_indices], color='red', s=100, marker='x', linewidths=3, label='Outliers')
axes[1].set_title('With Outliers', fontsize=12)
axes[1].legend()
axes[1].grid(True, alpha=0.3)

# Cleaned
axes[2].scatter(x_clean, y_clean, alpha=0.6, color='green')
axes[2].set_title('Cleaned Data', fontsize=12)
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Summary

In this notebook, we learned:

- How to visualize raw data
- How to detect and remove outliers
- How to normalize data using different methods

These preprocessing steps are essential before applying interpolation or approximation methods.
