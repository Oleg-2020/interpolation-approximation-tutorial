# Example 2: Interpolation Methods

Demonstrates linear, cubic spline, and polynomial interpolation techniques.

```python
import numpy as np
import matplotlib.pyplot as plt
from mathdoc import DataLoader, Interpolator
```

## 2.1 Load Data

```python
loader = DataLoader(random_seed=42)
x, y = loader.load_synthetic(n_points=15, noise=0.02, function='sine')

print(f'Data points: {len(x)}')
print(f'x range: [{x.min():.3f}, {x.max():.3f}]')
print(f'y range: [{y.min():.3f}, {y.max():.3f}]')
```

## 2.2 Linear Interpolation

```python
interp = Interpolator(x, y)

# Evaluate at dense grid
x_dense = np.linspace(x.min(), x.max(), 200)
y_linear = interp.linear(x_dense)

# Plot
plt.figure(figsize=(10, 5))
plt.plot(x_dense, y_linear, 'b-', linewidth=2, label='Linear interpolation')
plt.scatter(x, y, color='red', s=50, zorder=5, label='Data points')
plt.xlabel('x', fontsize=11)
plt.ylabel('y', fontsize=11)
plt.title('Linear Interpolation', fontsize=13, fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 2.3 Cubic Spline Interpolation

```python
# Cubic spline with dense evaluation points
y_spline = interp.cubic_spline(n_points=500)
x_spline = np.linspace(x.min(), x.max(), 500)

plt.figure(figsize=(10, 5))
plt.plot(x_spline, y_spline, 'g-', linewidth=2, label='Cubic spline')
plt.scatter(x, y, color='red', s=50, zorder=5, label='Data points')
plt.xlabel('x', fontsize=11)
plt.ylabel('y', fontsize=11)
plt.title('Cubic Spline Interpolation', fontsize=13, fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 2.4 Polynomial Interpolation

```python
# Polynomial interpolation (degree = n_points - 1)
degree = min(len(x) - 1, 8)  # limit degree
y_poly = interp.polynomial(degree=degree, x_new=x_dense)

plt.figure(figsize=(10, 5))
plt.plot(x_dense, y_poly, 'orange', linewidth=2, label=f'Polynomial (degree {degree})')
plt.scatter(x, y, color='red', s=50, zorder=5, label='Data points')
plt.xlabel('x', fontsize=11)
plt.ylabel('y', fontsize=11)
plt.title('Polynomial Interpolation', fontsize=13, fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 2.5 Comparison of Methods

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Linear
y_linear = interp.linear(x_dense)
axes[0, 0].plot(x_dense, y_linear, 'b-', linewidth=2)
axes[0, 0].scatter(x, y, color='red', s=50)
axes[0, 0].set_title('Linear', fontsize=12, fontweight='bold')
axes[0, 0].grid(True, alpha=0.3)

# Cubic spline
y_spline = interp.cubic_spline(x_new=x_dense)
axes[0, 1].plot(x_dense, y_spline, 'g-', linewidth=2)
axes[0, 1].scatter(x, y, color='red', s=50)
axes[0, 1].set_title('Cubic Spline', fontsize=12, fontweight='bold')
axes[0, 1].grid(True, alpha=0.3)

# Polynomial
y_poly = interp.polynomial(degree=8, x_new=x_dense)
axes[1, 0].plot(x_dense, y_poly, 'orange', linewidth=2)
axes[1, 0].scatter(x, y, color='red', s=50)
axes[1, 0].set_title('Polynomial (deg 8)', fontsize=12, fontweight='bold')
axes[1, 0].grid(True, alpha=0.3)

# All together
axes[1, 1].plot(x_dense, y_linear, 'b-', linewidth=1.5, alpha=0.7, label='Linear')
axes[1, 1].plot(x_dense, y_spline, 'g-', linewidth=1.5, alpha=0.7, label='Cubic Spline')
axes[1, 1].plot(x_dense, y_poly, 'orange', linewidth=1.5, alpha=0.7, label='Polynomial')
axes[1, 1].scatter(x, y, color='red', s=50, zorder=5)
axes[1, 1].set_title('Comparison', fontsize=12, fontweight='bold')
axes[1, 1].legend()
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Summary

- **Linear**: Fast, simple, but not smooth
- **Cubic Spline**: Smooth, natural-looking curves, widely used
- **Polynomial**: Can oscillate with high degrees (Runge's phenomenon)
